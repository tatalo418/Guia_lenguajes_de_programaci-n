```python
import pandas as pd
from openpyxl import Workbook

wb = Workbook()

# -------------------------
# HOJA 1: INSTRUCCIONES
# -------------------------
ws1 = wb.active
ws1.title = "Instrucciones"
ws1["A1"] = "Prueba Excel – Coordinador de Costos"
ws1["A3"] = "Instrucciones:"
ws1["A4"] = "1. Completar los ejercicios en las hojas siguientes."
ws1["A5"] = "2. No eliminar hojas ni modificar los encabezados."
ws1["A7"] = "Contenido:"
ws1["A8"] = "- Hoja 'Costos': Análisis de costos unitarios"
ws1["A9"] = "- Hoja 'Variaciones': Cálculo de variaciones"
ws1["A10"] = "- Hoja 'Dashboard': Crear un mini dashboard con gráficos"

# -------------------------
# HOJA 2: COSTOS
# -------------------------
ws2 = wb.create_sheet("Costos")
ws2.append(["Producto", "Costo Material", "Costo Mano Obra", "Costo Indirecto", "Total Costo Unitario"])
data = [
    ["Producto A", 12.5, 8.3, 4.2],
    ["Producto B", 9.8, 7.1, 3.6],
    ["Producto C", 15.2, 9.9, 5.1],
]
for row in data:
    ws2.append(row + [""])  # espacio para que el postulante calcule el costo total

# -------------------------
# HOJA 3: VARIACIONES
# -------------------------
ws3 = wb.create_sheet("Variaciones")
ws3.append(["Mes", "Costo Presupuestado", "Costo Real", "Variación"])
data2 = [
    ["Enero", 50000, 52000],
    ["Febrero", 48000, 47000],
    ["Marzo", 51000, 53000],
]
for row in data2:
    ws3.append(row + [""])  # espacio para que calcule variación

# -------------------------
# HOJA 4: DASHBOARD
# -------------------------
ws4 = wb.create_sheet("Dashboard")
ws4["A1"] = "Crear un mini-dashboard con:"
ws4["A2"] = "- Gráfico de variación mensual"
ws4["A3"] = "- Tabla resumen de costos unitarios"
ws4["A5"] = "Use las hojas 'Costos' y 'Variaciones' como fuente de datos."

# -------------------------
# GUARDAR ARCHIVO
# -------------------------
file_path = "Prueba_Coordinador_Costos.xlsx"
wb.save(file_path)

file_path
```
